```python
import os
import pandas as pd
import joblib
import shap
import matplotlib.pyplot as plt


BASE_DIR = os.getcwd()  # Current notebook directory
DATA_PATH = os.path.join(BASE_DIR, "../data/heart_disease_cleaned.csv")
MODEL_PATH = os.path.join(BASE_DIR, "../models/final_model.pkl")
SCALER_PATH = os.path.join(BASE_DIR, "../models/scaler.pkl")
IMAGE_DIR = os.path.join(BASE_DIR, "images")
os.makedirs(IMAGE_DIR, exist_ok=True)

df = pd.read_csv(DATA_PATH)
X = df.drop(columns=["target"])
y = df["target"]

model = joblib.load(MODEL_PATH)
scaler = joblib.load(SCALER_PATH)

# Scale features
X_scaled = scaler.transform(X)
X_scaled = pd.DataFrame(X_scaled, columns=X.columns)


# ------------------------------
explainer = shap.TreeExplainer(model)
shap_values = explainer(X_scaled)  # Use new SHAP API (v0.40+)

#  GLOBAL FEATURE IMPORTANCE
print("Generating global feature importance plots...")

# Bar plot (feature importance)
plt.figure(figsize=(10,6))
shap.summary_plot(shap_values.values, X_scaled, plot_type="bar", show=False)
plt.tight_layout()
plt.savefig(os.path.join(IMAGE_DIR, "shap_global_bar.png"))
plt.close()

# Dot plot (impact of features)
plt.figure(figsize=(12,8))
shap.summary_plot(shap_values.values, X_scaled, plot_type="dot", show=False)
plt.tight_layout()
plt.savefig(os.path.join(IMAGE_DIR, "shap_global_dot.png"))
plt.close()

print("Global SHAP plots saved in 'images/' folder.")

print("All SHAP plots saved successfully in 'images/' folder!")
```
